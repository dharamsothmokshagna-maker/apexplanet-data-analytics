# JOINS

Q. WHY WE USE JOIN

A. IN REAL DATABASES, DATA IS USUALLY STORED IN MULTIPLE TABLES 

HERE WE ARE USING SAMPLESUPERSTORE DATASET WE ONLY HAVE ONE TABLE FOR JOIN WE NEED 2 TABLES SO LET'S CREATE 2 TABLES

HERE WE CREATED TABLE 1: CUSTOMERS TABLE

```python
customers = df[['Customer ID', 'Customer Name', 'Segment']].drop_duplicates()

customers.to_sql('customers', conn, if_exists='replace', index=False)
```

# INNER JOIN

HERE IT SHOWS MATCHING RECORDS FROM BOTH TABLE1 AND TABLE2 TABLES

HERE LET ME EXPLAIN SOMETHING WHICH WE NEED TO UNDERSTAND

c --> CUSTOMERS TABLE ALIES

o --> ORDERS TABLE ALIES

IT JUST SHOWS MATCHING ROWS AND CUSTOMER ID WHICH IS SIMILAR TO EACH OTHER

```python
query = """
SELECT
c.[Customer Name],
c.Segment,
o.Sales,
o.Profit
FROM customers c
INNER JOIN orders o
ON c.[Customer ID] = o.[Customer ID]
LIMIT 10
"""

pd.read_sql(query, conn)
```

# LEFT JOIN

IT SHOWS ALL CUSTOMERS, EVEN IF THEY HAVE NO ORDERS

```python
query = """
SELECT
c.[Customer Name],
c.Segment,
o.Sales
FROM customers c
LEFT JOIN orders o
ON c.[Customer ID] = o.[Customer ID]
LIMIT 10
"""

pd.read_sql(query, conn)
```

# RIGHT JOIN

HERE WE NEED TO KNOW ABOUT SQLITE DOESN'T SUPPORT RIGHT JOIN DIRECTLY , WE USE THIS CODE

```python
query = """
SELECT
c.[Customer Name],
o.Sales
FROM orders o
LEFT JOIN customers c
ON o.[Customer ID] = c.[Customer ID]
LIMIT 10
"""

pd.read_sql(query, conn)
```

# SUBQUERY

SUBQUERY IS A QUERY INSIDE ANOTHER QUERY

```python
# HERE WE FIND ORDERS ABOVE AVERAGE SALES
query="""
SELECT *
FROM sales
WHERE Sales >
(
    SELECT AVG(Sales)
    FROM sales
    )
    LIMIT 10
    """
pd.read_sql(query,conn)
```

```python
# HERE WE FIND PRODUCTS WITH HIGHEST PROFIT
query="""
SELECT *
FROM sales
WHERE Profit >
(
    SELECT MAX(Profit)
    FROM sales
    )
    """
pd.read_sql(query,conn)
```

```python
# HERE WE FIND CUSTOMERS WHOSE SALES ARE ABOVE AVERAGE
query="""
SELECT *
FROM sales
WHERE Sales >
(
    SELECT AVG(Sales)
    FROM sales
    )
    ORDER BY Sales DESC
    LIMIT 10
    """
pd.read_sql(query,conn)
```

```python
# HERE WE FINDED STATES HAVING SALES GREATER THAN AVERAGE STATE SALES
query="""
SELECT State,
SUM(Sales) AS Total_Sales
FROM sales
GROUP BY State
HAVING SUM(Sales) >
(
    SELECT AVG(Sales)
    FROM sales
    )
    """
pd.read_sql(query,conn)
```

# CTE(WITH CLAUSE)

A CTE (COMMON TABLE EXPRESSION) CREATES A TEMPORARY RESULT SET THAT CAN BE USED IN THE MAIN QUERY

```python
# HERE WE FIND CATEGORY SALES ANALYSIS
query = """
WITH CategorySales AS
(
    SELECT Category,
           SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Category
)

SELECT *
FROM CategorySales
ORDER BY Total_Sales DESC
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND CATEGORY PROFIT ANALYSIS
query = """
WITH CategoryProfit AS
(
    SELECT Category,
           SUM(Profit) AS Total_Profit
    FROM sales
    GROUP BY Category
)

SELECT *
FROM CategoryProfit
ORDER BY Total_Profit DESC
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND STATE SALES ANALYSIS
query = """
WITH StateSales AS
(
    SELECT State,
           SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY State
)

SELECT *
FROM StateSales
WHERE Total_Sales > 50000
ORDER BY Total_Sales DESC
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND STATE PROFIT ANALYSIS
query = """
WITH StateProfit AS
(
    SELECT State,
           SUM(Profit) AS Total_Profit
    FROM sales
    GROUP BY State
)

SELECT *
FROM StateProfit
ORDER BY Total_Profit DESC
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND TOP 10 CUSTOMERS SALES USING CTE
query = """
WITH CustomerSales AS
(
    SELECT [Customer Name],
           SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY [Customer Name]
)

SELECT *
FROM CustomerSales
ORDER BY Total_Sales DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND TOP 10 CUSTOMERS PROFIT USING CTE
query = """
WITH CustomerProfit AS
(
    SELECT [Customer Name],
           SUM(Profit) AS Total_Profit
    FROM sales
    GROUP BY [Customer Name]
)

SELECT *
FROM CustomerProfit
ORDER BY Total_Profit DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE WE FIND CATEGORY SALES AND PROFIT TOGETHER
query = """
WITH CategoryStats AS
(
    SELECT Category,
           SUM(Sales) AS Total_Sales,
           SUM(Profit) AS Total_Profit
    FROM sales
    GROUP BY Category
)

SELECT *
FROM CategoryStats
ORDER BY Total_Profit DESC
"""

pd.read_sql(query, conn)
```

# ROW()

WHY WE USE ROW_NUMBER ?

WE USE ROW_NUMBER BECAUSE IT GIVES A UNIQUE NUMBER TO EACH ROW

```python
# HERE IT SHOWS TOP SALES RANKING
query = """
SELECT
[Customer Name],
Sales,
ROW_NUMBER() OVER(
ORDER BY Sales DESC
) AS Row_Num
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE IT SHOWS TOP PROFITS RANKING
query = """
SELECT
[Customer Name],
Profit,
ROW_NUMBER() OVER(
ORDER BY Profit DESC
) AS Row_Num
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

HERE WHY WE USE PARTITION BY BECAUSE IT CREATES SEPERATES RANKINGS INSIDE EACH CETEGORY

FOR EX : TECHNOLOGY IT DECLARES RANK 1, RANK 2,......

```python
# HERE WE CAN SEE ROW NUM WITHIN EACH CATEGORY
query = """
SELECT
Category,
[Customer Name],
Sales,
ROW_NUMBER() OVER(
PARTITION BY Category
ORDER BY Sales DESC
) AS Category_Rank
FROM sales
LIMIT 20
"""

pd.read_sql(query, conn)
```

# LAG( )

WHY WE USE LAG( )?

WE USE LAG( ) BECAUSE IT ACCESSES THE PREVIOUS ROW'S VALUE

```python
# HERE WE CAN SEE PREVIOUS SALES VALUE 
query = """
SELECT
[Order Date],
Sales,
LAG(Sales,1) OVER(
ORDER BY [Order Date]
) AS Previous_Sales
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE WE CAN SEE PREVIOUS PROFIT VALUE 
query = """
SELECT
[Order Date],
Profit,
LAG(Profit,1) OVER(
ORDER BY [Order Date]
) AS Previous_Profit
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE BY USING LAG WE CAN SEE SALES DIFFERENCE
query = """
SELECT
[Order Date],
Sales,
Sales -
LAG(Sales,1) OVER(
ORDER BY [Order Date]
) AS Sales_Difference
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
# HERE BY USING LAG WE CAN SEE PROFIT DIFFERENCE
query = """
SELECT
[Order Date],
Profit,
Profit -
LAG(Profit,1) OVER(
ORDER BY [Order Date]
) AS Profit_Difference
FROM sales
LIMIT 10
"""

pd.read_sql(query, conn)
```

Q 1 . WHICH CATEGORY GENERATES THE HIGHEST SALES?

```python
query= """
SELECT Category,
SUM(Profit) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Profit DESC
"""
pd.read_sql(query,conn)
```

Q 3. WHAT ARE THE TOP 10 CUSTOMERS BY SALES?

Q 4. WHAT ARE THE TOP 5 PRODUCTS BY SALES?

Q 5. WHICH STATES GENERATE THE HIGHEST REVENUE?

Q 6. WHICH STATE GENERATE THE HIGHEST PROFIT?

Q 8. WHICH PRODUCTS ARE CAUSING LOSSES?

Q 9. MONTHLY SALES TREND?

Q 10. WHICH CUSTOMERS ARE HIGH,MEDIUM,AND LOW SPENDERS?

```python
query = """
SELECT
[Customer Name],
SUM(Sales) AS Total_Spend,
CASE
    WHEN SUM(Sales) >= 5000 THEN 'High Spender'
    WHEN SUM(Sales) >= 2000 THEN 'Medium Spender'
    ELSE 'Low Spender'
END AS Customer_Segment
FROM sales
GROUP BY [Customer Name]
ORDER BY Total_Spend DESC
"""

pd.read_sql(query, conn)
```

```python
import sqlite3
import pandas as pd

# Database Connection Function
def connect_db(db_name="superstore.db"):
    return sqlite3.connect(db_name)

# Execute SELECT Queries
def run_query(query):
    conn = connect_db()
    result = pd.read_sql(query, conn)
    conn.close()
    return result
```

```python
query= """
SELECT Category,
SUM(Sales) AS Total_Sales
FROM sales
GROUP BY Category
"""
run_query(query)
```
