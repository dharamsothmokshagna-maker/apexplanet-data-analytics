```python
#here it deletes any duplicate rows found.but we already got 0 ,it didnt need to remove anything
df=df.drop_duplicates()
```

```python
df.shape
```

```python
#it scans every single cell in our huge data tables
#it repaces our actual data values with true or false words
# false means cell is not empty(it contain real data like a name,date,or numbers)
# true means the cell is blank or empty
df.isnull()
```

```python
#instead of reading 1000 rows,it gives us a clean checklist.if a column shows 0,it means it is fully complete
df.isnull().sum()#these are the empty cells present in each column
```

```python
# fillna(0) it fills every blank cell in this column,and fill it with a 0
df['Postal Code']=df['Postal Code'].fillna(0)
```

```python
print(df['Postal Code'].isnull().sum())
```

```python
# here we are converting text format date into date format
df['Order Date'] = pd.to_datetime(df['Order Date'],format="mixed")
df['Ship Date'] = pd.to_datetime(df['Ship Date'],format="mixed")
```
