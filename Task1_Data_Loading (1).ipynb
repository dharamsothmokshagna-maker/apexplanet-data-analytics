```python
#Step 5 : Handle ouliers using IQR method
```

```python
# Outlier detection Q1(first quartile) and Q3(third quartile)
# q1 means 25% of data lies below 25
# q3 means 75% of data lies below 65
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

# Calculate IQR
IQR = Q3-Q1

lower= Q1-1.5*IQR
upper= Q3+1.5*IQR
```

```python
outliers = df[(df['Sales']<lower)  | (df['Sales']>upper)]
print(outliers.shape)
```
